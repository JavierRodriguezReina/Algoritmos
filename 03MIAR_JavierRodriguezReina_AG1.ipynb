## Divide y vencerás - Problema Torres de Hanói
- Parto de un problema y lo soluciono mediante la solución de subproblemas y unión de resultados
- En este caso concreto: H(n) = H(n-1) + 1 + H(n-1)
- Pensar lo anterior como... "Muevo todos los discos salvo uno a torre auxiliar, muevo el disco más grande a la torre final y muevo todos los discos restantes de la torre auxiliar a la final"
- Solamente resuelvo el problema de mover un disco (condición de parada de división) porque el resto se aplica el mismo algoritmo.
- Mover un disco es muy sencillo -> Solución directa -> Mover de torre de incio a torre de fin
- Ojo, el print del "if" es mover un disco entre torres (cualesquiera) y el print del "else" es el caso de mover el disco grande a torre final

````python
def torres_hannoi(numero_discos, inicio, fin):

    #print(f"torres_hannoi({numero_discos}, {inicio}, {fin})")

    # Si el número de discos es 1 -> Caso base, mover de inicio a fin
    if numero_discos == 1:
        #print(f"Caso base: Mover desde {inicio} a {fin} -- torres_hannoi({numero_discos}, {inicio}, {fin})")
        print(f"Mover desde {inicio} a {fin}")
    # Si el número de discos es mayor que 1 -> Mover resto de discos, mover disco único,
    # recuperar => Recursivamente
    else:
        torres_hannoi(numero_discos-1, inicio, 6-inicio-fin)
        #print(f"Caso central: Mover desde {inicio} a {fin} -- torres_hannoi({numero_discos}, {inicio}, {fin})")
        print(f"Mover desde {inicio} a {fin}")
        torres_hannoi(numero_discos-1, 6-inicio-fin, fin)

torres_hannoi(3,1,3)
````

## Técnicas voraces - Problema del cambio
- Para resolver el problema, elijo una solución "buena" sin pensar en consecuencias
- Sería como elegir algo "óptimo" localmente con la esperanza de que sea lo mejor en general
- No siempre funciona bien -> Si en el conjunto de las soluciones, dos están próximas entre sí y generan una solución mayor puede fallar (ej: monedas [1, 3, 4, 9])

````python
def cambio_monedas(sistema, cambio):

    print(f"Sistema: {sistema}")
    print(f"Cambio: {cambio}")

    # Inicialización de resultados
    resultado = [0] * len(sistema)

    # Para cada posible moneda, calculamos el número de monedas máximo que podemos dar
    # y lo restamos del cambio pendiente (mejor decisión local, que no tiene por qué
    # ser óptima
    for i in range(len(sistema)):
        total_monedas = cambio//sistema[i] # División con resultado entero
        resultado[i] = total_monedas
        cambio = cambio - resultado[i]*sistema[i]

    return resultado

# Sistema óptimo
sistema = [25, 10, 5, 1]
cambio = 27
resultado = cambio_monedas(sistema,cambio)
print(f"Resultado: {resultado}")

print("")

# Sistema NO óptimo
sistema = [11, 5, 1]
cambio = 15
resultado = cambio_monedas(sistema,cambio)
print(f"Resultado: {resultado}")
````

## Backtracking - Problema de las N reinas
- Se divide el problema en etapas
- En cada etapa se construye una solución parcial
- Se comprueba si esa solución parcial sigue siendo válida
- Si es válida, se continúa con la siguiente etapa
- Si no es válida, se descarta y se prueba otra alternativa
- Si ninguna alternativa funciona, se retrocede a la etapa anterior para cambiar decisiones

````python
# Función que comprueba si por columnas hay un error en la solución
# etapa(i) = etapa (j) -> Error
# También comprueba si por diagonales hay un error en la solución
# (abs(i-j) = abs(xi-xj)) -> Error
def conflicto(solucion, etapa):
    for i in range(etapa+1):
        if solucion.count(solucion[i]) > 1:
            return True
        for j in range(etapa):
            if (i != j):
                if (abs(i-j) == abs(solucion[i]-solucion[j])):
                    return True
    else:
        return False
````

````python
# Búsqueda de una solución
def colocar_reinas(numero_reinas, solucion = None, etapa = 0):

    if solucion is None:
        solucion = numero_reinas*[0]

    # Caso base: si he colocado todas las reinas (he completado todas las etapas)
    # etapa comienza en 0 y llega hasta numero_reinas
    if numero_reinas == etapa:
        #print(f"({numero_reinas}, {solucion}, {etapa})")
        print(f"{solucion}")
        return solucion

    else:
        # Recorro todas las columnas posibles para la reina en la fila (etapa)
        for i in range(1,numero_reinas+1):
            solucion[etapa] = i
            #print(f"({numero_reinas}, {solucion}, {etapa})")

            # Compruebo si la colocación actual no entra en conflicto
            # con reinas anteriores (columnas y diagonales)
            if not (conflicto(solucion, etapa)):
                # Si es válida, intento resolver el resto del tablero
                resultado = colocar_reinas(numero_reinas, solucion, etapa + 1)

                # Si desde aquí se encuentra una solución válida,
                # la propago hacia arriba para detener la búsqueda
                # En otras palabras, ya tengo resultado, no necesito
                # seguir probando, lo devuelvo
                if resultado is not None:
                    return resultado

        # Limpio la posición de la reina si todos los estados fallan (vuelta atrás)
        solucion[etapa] = 0

        # Si ninguna columna funciona en esta etapa, esta rama no tiene solución
        #return None

solucion = colocar_reinas(4)
````

````python
def es_prometedora(solucion,etapa):
    for i in range(etapa+1):
        if solucion.count(solucion[i]) > 1:
            return False
        for j in range(i+1,etapa+1):
            if abs(i-j)==abs(solucion[i]-solucion[j]):
                return False
    return True


def reinas(N,solucion=[],etapa=0):
    if len(solucion) == 0:
        solucion = [0 for i in range(N)]
    for i in range(1,N+1):
        solucion[etapa] = i
        if es_prometedora(solucion,etapa):
            if etapa == N-1:
                print(solucion)
            else:
                reinas(N,solucion,etapa+1)
        else:
            None

        solucion[etapa] = 0

reinas(4)
````

````python
def torres_hanoi(n, origen, destino, auxiliar):
  if n == 1:
    print(f"Mover disco 1 de {origen} a {destino}")
    return

  # Paso 1: Mover n-1 discos de origen a auxiliar usando destino
  torres_hanoi(n - 1, origen, auxiliar, destino)

  # Paso 2: Mover el disco n (el más grande) de origen a destino
  print(f"Mover disco {n} de {origen} a {destino}")

  # Paso 3: Mover n-1 discos de auxiliar a destino usando origen
  torres_hanoi(n - 1, auxiliar, destino, origen)

torres_hanoi(3, 'A', 'C', 'B')
````

````python
import numpy as np

num_puntos = 500

puntos_1d = np.random.uniform(0, 10000, num_puntos)            # Lista/Array 1D
puntos_2d = np.random.uniform(0, 100, size=(num_puntos, 2))    # Lista/Array 2D (X, Y)
puntos_3d = np.random.uniform(0, 100, size=(num_puntos, 3))    # Lista/Array 3D (X, Y, Z)


# 2. Creamos y ORDENAMOS los puntos a partir de las listas anteriores
# De esta forma contienen exactamente los mismos datos pero preparados para Divide y Vencerás
puntos_1d_sorted = np.sort(puntos_1d)
puntos_2d_sorted = np.array(sorted(puntos_2d.tolist(), key=lambda p: p[0]))
puntos_3d_sorted = np.array(sorted(puntos_3d.tolist(), key=lambda p: p[0]))
````

### Fuerza Bruta 1D

````python
def par_mas_cercano_1d(puntos):
    distancia_minima = float('inf')
    punto1_optimo = None
    punto2_optimo = None
    n = len(puntos)

    for i in range(n):
        for j in range(i + 1, n):
            distancia = abs(puntos[i] - puntos[j])

            if distancia < distancia_minima:
                distancia_minima = distancia
                punto1_optimo = puntos[i]
                punto2_optimo = puntos[j]

    return (punto1_optimo, punto2_optimo), distancia_minima

# Ejecución 1D
pareja_1d, dist_1d = par_mas_cercano_1d(puntos_1d)
print(f"1D -> Pareja más cercana: {pareja_1d} | Distancia: {dist_1d:.4f}")
````


### Complejidad Fuerza Bruta 1D

La línea clave que se repite es:


```
 distancia = abs(puntos_1d[i] - puntos_1d[j])
```

 Depende de los bucles anidados:

```

  • i = 0   →  j va de 1 a n-1     →  n - 1  veces
  • i = 1   →  j va de 2 a n-1     →  n - 2  veces
  • ...
  • i = n-2 →  j vale n-1          →  1      veces
  • i = n-1 →  j ya no tiene rango →  0      veces
```



Suma total de ejecuciones:


```
  S = (n - 1) + (n - 2) + ... + 1 + 0
  S = [ n * (n - 1) ] / 2 = (n² - n) / 2
```


Aplicando Notación O:

• Eliminamos la constante /2.

• Eliminamos el término de menor grado (-n) porque para n = 1,000,000, n² (un billón) domina por completo a n (un millón).

  
Complejidad = O(n²)

Se podría mejorar mediante divide y vencerás, donde se  divide el conjunto a la mitad de forma recursiva de tal forma que al combinar las partes, solo necesita comparar el elemento más a la derecha del subconjunto izquierdo con el más a la izquierda del derecho. Esto reduce radicalmente la complejidad a O(n²).

### Divide y Vencerás 1D



````python
def par_cercano_1d_dv(puntos):
    n = len(puntos)
    if n <= 3:
        min_dist = float('inf')
        par = None
        for i in range(n):
            for j in range(i + 1, n):
                d = abs(puntos[i] - puntos[j])
                if d < min_dist:
                    min_dist, par = d, (puntos[i], puntos[j])
        return par, min_dist

    # Partimos la lista a la mitad
    mitad = n // 2
    izq = puntos[:mitad]
    der = puntos[mitad:]

    # Resolvemnos recursivamente cada mitad
    par_izq, dist_izq = par_cercano_1d_dv(izq)
    par_der, dist_der = par_cercano_1d_dv(der)

    # Nos quedamos con el mínimo de ambos lados
    if dist_izq < dist_der:
        mejor_par, mejor_dist = par_izq, dist_izq
    else:
        mejor_par, mejor_dist = par_der, dist_der

    # Comparamos el punto frontera (el último de la izq con el primero de la der)
    dist_frontera = abs(izq[-1] - der[0])
    if dist_frontera < mejor_dist:
        mejor_dist = dist_frontera
        mejor_par = (izq[-1], der[0])

    return mejor_par, mejor_dist

# Ejecución 1D
par_1d, d_1d = par_cercano_1d_dv(puntos_1d_sorted)
print(f"1D -> Pareja más cercana: {par_1d} | Distancia: {d_1d:.4f}")
````

### Complejidad Divide y Vencerás 1D

La línea clave que se repite en la fase de combinación es:

`  distancia_frontera = abs(izq[-1] - der[0])`

Al dividir el problema a la mitad de forma recursiva, el comportamiento se rige por la ecuación de recurrencia:

`  T(n) = 2T(n/2) + O(1)`

Donde:


  • 2T(n/2) es el tiempo invertido en resolver las dos mitades.
  
  • O(1) es el tiempo que toma combinar los resultados (ya que solo comparamos un par frontera: el último de la izquierda con el primero de la derecha).



Aplicando Notación O:



• El proceso de división genera un árbol de recursión de altura log₂(n).

• En cada nivel del árbol se realiza un trabajo constante de combinación.

• El coste del ordenamiento inicial obligatorio es de O(n log n).





Complejidad = O(n log n)

### Fuerza Bruta 2D

````python
def par_mas_cercano_2d(puntos):
    distancia_minima = float('inf')
    punto1_optimo = None
    punto2_optimo = None
    n = len(puntos)

    for i in range(n):
        for j in range(i + 1, n):
            # Distancia en 2D
            distancia = np.linalg.norm(puntos[i] - puntos[j])

            if distancia < distancia_minima:
                distancia_minima = distancia
                punto1_optimo = puntos[i]
                punto2_optimo = puntos[j]

    return (punto1_optimo, punto2_optimo), distancia_minima

# Ejecución 2D
pareja_2d, dist_2d = par_mas_cercano_2d(puntos_2d)
print(f"2D -> Pareja más cercana: {tuple(pareja_2d)} | Distancia: {dist_2d:.4f}")
````

### Complejidad Fuerza Bruta 2D

La estructura de los bucles es exactamente la misma que en 1D, por lo que el número de comparaciones sigue el mismo patrón. El único cambio es la manera de calcular la distancia, que pasa a incluir dos coordenadas (X e Y), pero esto no afecta a la complejidad del algoritmo.

La línea clave que se repite es:

```
distancia = np.linalg.norm(puntos_2d[i] - puntos_2d[j])
```

Depende de los bucles anidados:

```
  • i = 0   →  j va de 1 a n-1     →  n - 1  veces
  • i = 1   →  j va de 2 a n-1     →  n - 2  veces
  • ...
  • i = n-2 →  j vale n-1          →  1      veces
  • i = n-1 →  j ya no tiene rango →  0      veces
```
Suma total de ejecuciones:

```
  S = (n - 1) + (n - 2) + ... + 1 + 0
  S = [ n * (n - 1) ] / 2 = (n² - n) / 2
```

Aplicando Notación O:




• Aunque internamente `np.linalg.norm` calcula dos coordenadas √( (x2-x1)² + (y2-y1)² ), el número de comparaciones de pares de
puntos sigue la misma serie matemática.

• Eliminamos la constante /2.

• Eliminamos el término de menor grado (-n) porque para n = 1,000,000, n² (un billón) domina por completo a n (un millón).



Complejidad = O(n²)

Se podría mejorar mediante divide y venceras dividiendo el plano con una línea vertical imaginaria y resolviendo cada lado. El reto está en la zona de combinación, donde busca puntos cercanos a la línea divisoria usando una "franja" estrecha, lo que también reduce el tiempo a O(n²).


### Divide y Vencerás 2D

````python
def par_cercano_2d_dv(puntos):
    n = len(puntos)
    if n <= 3:
        min_dist = float('inf')
        par = None
        for i in range(n):
            for j in range(i + 1, n):
                d = np.linalg.norm(puntos[i] - puntos[j])
                if d < min_dist:
                    min_dist, par = d, (puntos[i], puntos[j])
        return par, min_dist

    # Dividimos
    mitad = n // 2
    linea_x = puntos[mitad][0] # Coordenada X que divide el plano

    # Resolvemnos recursivamente cada mitad
    (par_izq, dist_izq) = par_cercano_2d_dv(puntos[:mitad])
    (par_der, dist_der) = par_cercano_2d_dv(puntos[mitad:])

    if dist_izq < dist_der:
        mejor_par, mejor_dist = par_izq, dist_izq
    else:
        mejor_par, mejor_dist = par_der, dist_der

    # Filtrar puntos dentro de la franja central (eje X)
    franja = [p for p in puntos if abs(p[0] - linea_x) < mejor_dist]
    # Ordenamos la franja por el eje Y para comparar solo vecinos cercanos
    franja.sort(key=lambda p: p[1])

    # Buscamos en la franja
    for i in range(len(franja)):
        for j in range(i + 1, len(franja)):
            if (franja[j][1] - franja[i][1]) >= mejor_dist:
                break
            d = np.linalg.norm(franja[i] - franja[j])
            if d < mejor_dist:
                mejor_dist = d
                mejor_par = (franja[i], franja[j])

    return mejor_par, mejor_dist

# Ejecución 2D
par_2d, d_2d = par_cercano_2d_dv(puntos_2d_sorted)
print(f"2D -> Pareja más cercana: {par_2d} | Distancia: {d_2d:.4f}")
````

Complejidad Divide y Vencerás 2D

La complejidad se mantiene en el mismo orden de magnitud que en 1D. El único cambio es la manera de calcular la distancia en el plano y que el proceso de combinación requiere analizar una franja vertical en el eje Y, pero esto no afecta la curva de crecimiento asintótico.

La línea clave que se repite en la fase de combinación dentro de la franja es:

`  d = np.linalg.norm(franja[i] - franja[j])`

La ecuación de recurrencia para este caso es:

`  T(n) = 2T(n/2) + O(n)`

Donde:


  • 2T(n/2) representa resolver los subproblemas de cada mitad.

  • O(n) representa el tiempo que toma filtrar y ordenar los puntos dentro de la franja central en cada paso. Aunque se usa un doble bucle para comparar los puntos de la franja, geométricamente se demuestra que cada punto se compara con un número máximo constante de vecinos (máximo 7), por lo que esa búsqueda interna es lineal O(n).


Aplicando Notación O:


• El coste de la combinación O(n) repetido a lo largo de los log₂(n) niveles de división resulta en un tiempo total de O(n log n).




Complejidad = O(n log n)

````python
def par_cercano_3d_dv(puntos):
    n = len(puntos)
    if n <= 3: # Caso base
        min_dist = float('inf')
        par = None
        for i in range(n):
            for j in range(i + 1, n):
                d = np.linalg.norm(puntos[i] - puntos[j])
                if d < min_dist:
                    min_dist, par = d, (puntos[i], puntos[j])
        return par, min_dist

    # 1. DIVIDIR
    mitad = n // 2
    plano_x = puntos[mitad][0] # Plano X que divide el espacio

    # 2. VENCER
    (par_izq, dist_izq) = par_cercano_3d_dv(puntos[:mitad])
    (par_der, dist_der) = par_cercano_3d_dv(puntos[mitad:])

    if dist_izq < dist_der:
        mejor_par, mejor_dist = par_izq, dist_izq
    else:
        mejor_par, mejor_dist = par_der, dist_der

    # 3. COMBINAR: Filtrar puntos dentro de la franja en el espacio 3D
    franja = [p for p in puntos if abs(p[0] - plano_x) < mejor_dist]
    franja.sort(key=lambda p: p[1]) # Ordenamos por el eje Y para la comparación

    # Buscamos en el volumen de la franja
    for i in range(len(franja)):
        for j in range(i + 1, len(franja)):
            if (franja[j][1] - franja[i][1]) >= mejor_dist:
                break
            d = np.linalg.norm(franja[i] - franja[j])
            if d < mejor_dist:
                mejor_dist = d
                mejor_par = (franja[i], franja[j])

    return mejor_par, mejor_dist

# Ejecución 3D
par_3d, d_3d = par_cercano_3d_dv(puntos_3d_sorted)
print(f"3D -> Pareja más cercana: {par_3d} | Distancia: {d_3d:.4f}")
````

### Complejidad Divide y Vencerás 3D

Al igual que en 2D, la complejidad sigue siendo la misma porque la estructura de divisiones sucesivas no varía. El único cambio es la manera de calcular la distancia (añadiendo el eje Z) y que la combinación opera sobre una franja volumétrica en lugar de una plana, pero controlando que las comparaciones internas no se disparen.

La línea clave que se repite en la combinación es:

`  d = np.linalg.norm(franja[i] - franja[j])`

La ecuación de recurrencia se mantiene como:

`  T(n) = 2T(n/2) + O(n)`

Donde:

  • 2T(n/2) es la división del espacio tridimensional mediante planos.

  • O(n) es el filtrado y escaneo de la caja o franja central. Al igual que en el plano bidimensional, la densidad geométrica de puntos que pueden estar cerca de la frontera sin ser ya descartados está limitada por una constante fija, impidiendo que el bucle interno se vuelva cuadrático.




Aplicando la Notación O:

• El crecimiento del algoritmo al aumentar 'n' sigue estrictamente la misma progresión lineal-logarítmica gracias a la estrategia de particionado eficiente de datos.
